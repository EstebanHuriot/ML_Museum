# Logistic Regression

```python
import numpy as np
import matplotlib.pyplot as plt 

np.random.seed(42)
```

"Just like a Linear Regression model, a Logistic Regression model computes a weighted sum of the input features (plus a bias term), but instead of outputting the results 
directly like the Linear Regression model does, it outputs the logistic of this result". Aurélien Géron, Hands-On Machine Learning with Scikit-Learn, Keras, and TensorFlow
</br>
</br>
The logistic, also noted σ() is a sigmoid function that outputs a value between 0 and 1:
</br>
$$
\sigma(t) = \frac{1}{1 + e^{-t}}
$$
</br>
So when the estimated probability is >= 50% the logistic regression outputs a 1, else it is a 0.

Also, the vectorized form looks like this: p_pred = σ(X θ).

# Generating a synthetic dataset

```python
n_sample = 100

X = np.random.randint(1, 100, size=(n_sample, 1))
noise = np.random.normal(0, 1, size=(n_sample, 1))

z = 0.12 * X - 6 + noise

# sigmoid
p = 1 / (1 + np.exp(-z))

# random 0 or 1
y = np.random.binomial(1, p).ravel()

plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Synthetic logistic dataset")
plt.show()
```

# Training

If the real class is 0, we want our model to predict something close to 0, a contrario if the real class is 1 we want the model to predict something close to 1.
</br>
For a single X we can use this cost function c():
</br>
$$
c(\theta) =
\begin{cases}
-\log(p) & \text{if } y = 1 \\
-\log(1 - p) & \text{if } y = 0
\end{cases}
$$
